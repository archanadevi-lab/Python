# **16. Check if a number is a palindrome**

Write a program to check if a given number is a palindrome. A palindrome number is a number that remains the same when its digits are reversed (e.g., 121, 343).

```python
a=int(input())
d=a
c=0
while a>0:
  b=a%10
  c=c*10+b
  a=a//10
if c==d:
    print("True")
else:
    print("Not a palindrome")
```

# **18. Collatz Conjecture: Generate a sequence until it reaches 1**

The Collatz conjecture states that if you start with any positive integer n, and if n is even, divide it by 2; if n is odd, multiply it by 3 and add 1. Repeat the process. The sequence will always eventually reach 1. Write a program to print this sequence for a given number.

```python
num=int(input())
print(num,end="")

while num!=1:
  if num%2==0:
     num=num//2
  else :
    num=num*3+1
  print(f",{num}",end="")
```

# **19. Armstrong Number Check**

 Write a program to check if a number is an Armstrong number. An Armstrong number (for a 3-digit number) is an integer such that the sum of the cubes of its digits is equal to the number itself (e.g., 153 = 1^3 + 5^3 + 3^3).

```python
a=int(input())
b=str(a)
power=len(b)
res=0

for i in b:
  res+=int(i)**power

if res==a:
  print(f"{a} is an Armstrong number")

else:
  print(f"{a} is not an Armstrong Number")

```

# **24. Hollow square pattern**

Print a 5*5 square of stars where the middle is empty, leaving only the border.

```python
n=int(input())
for i in range(0,n):
  for j in range(0,n):
    if i==0 or i==n-1 or j==0 or j==n-1:
      print("*",end=" ")
    else:
      print(" ",end=" ")
  print()

```

# **35. Perfect number check**

Write a program to check if a number is a “Perfect Number.” A perfect number is a positive integer that is equal to the sum of its proper divisors (excluding the number itself). For example, 6 is perfect because 1 + 2 + 3 = 6.

```python
num=15
divisor_Sum=0

for i in range(1,num//2+1):
  if num%i==0:
    divisor_Sum+=i


if divisor_Sum==num:
  print(f"The number{num} is a perfect number")

else:
  print(f"The number {num} is not a perfect number")
```

# **36. Binary to decimal conversion using loop**

Manually convert a binary string (e.g., "1101") into its decimal integer equivalent using a loop. Do not use int(binary, 2).

```python
a=str(1101)
pow=0
res=0
for i in a[::-1]:
  res+=(int(i)*(2**pow))
  pow+=1
print(f"The decimal integer of binary {a} is {res}")
```

# **40. Nested list search (2D matrix coordinates)**

 Given a 2D list (matrix), find the row and column index of a target value.

```python
matrix = [[10, 20], [30, 40], [50, 60]]
target = 30

for row_id,row in enumerate(matrix):
  for col_id,col in enumerate(row):
    if target==col:
      print(row_id,col_id)
```
