```python
# For loop- is used when we know the no. of Iterations.
# While loop- iterate until a condition becomes false
#Looping a string and indexing
s="hello"
for i in range(0,len(s)):
  print(s[i])
```

```python
#Looping a string
s="hello"
for i in s:
  print(i)
```

```python
#Looping a dictionary
d={1:100, 2:200, 3:300}
for i in d:
  print(i, end ='\t')
  print(i,d[i])
```

```python
#While loop -String
s="This is a while loop"
inc=0
while inc<len(s) :
  print(s[inc])
  print()
  inc+=1
print("Iteration finished")
```

```python
# Break -- Comes out of the loop
# Continue - Comes out of current iteration
```

```python
#Break
lst=range(10)
for i in lst:
  print(i)
  if i==6:
    break
    print(1)
```

```python
#Continue
for i in range(10):
    if i%3:
           continue
    else:
      print(i)
```
