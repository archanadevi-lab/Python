```python
#Calculator for performing Arithmetic Calculations
op=int(input('''Enter an operation:
              \t \t 1.Addition
              \t \t 2.Subtraction
              \t \t 3.Multiplication
              \t \t 4.Float division
              \t \t 5.Floor division
              \t \t 6.Modulo             '''))
a=int(input("Enter a value:"))
b=int(input("Enter b value:"))

if op==1 :
 print(a+b)
elif op==2 :
  print(a-b)
elif op==3 :
  print(a*b)
elif op==4 :
  print(a/b)
elif op==5:
   print(a//b)
elif op==6 :
  print(a%b)
else :
 print("please specify the given operation")
```

```python
# Check for leap year
year= int(input("Enter an year to check if it is a leap year:"))
if year%4==0 and year%100!=0:
  print(f'{year} is a leap year')
elif year%400==0:
  print(f'{year} is a leap year')
else:
  print(f'{year} is not a leap year')
```
